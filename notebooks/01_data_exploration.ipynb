Это исследовательский блокнот, а не финальный pipeline. Он использует небольшой
фиксированный набор из 2 452 запросов и benchmark-корпус для быстрого сравнения идей.
Финальная честная валидация строится отдельно в `02_candidate_pipeline.ipynb`.

## 3. Бейзлайн: TF-IDF по заголовкам

Начинаем только с `search_query` и `item_title_raw`


```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.neighbors import NearestNeighbors
```

```python
item_texts = benchmark_items["item_title_raw"].fillna("").astype(str)
query_texts = local_queries["search_query"].fillna("").astype(str)
```

```python
vectorizer = TfidfVectorizer(
    lowercase=True,
)
item_matrix = vectorizer.fit_transform(item_texts)
query_matrix = vectorizer.transform(query_texts)
```

```python
item_matrix.shape, query_matrix.shape
```

```python
search_index = NearestNeighbors(
    n_neighbors=50,
    metric="cosine",
    algorithm="brute",
    n_jobs=-1,
)

search_index.fit(item_matrix)
distances, indices = search_index.kneighbors(query_matrix)
indices.shape, distances.shape
```

```python
item_ids = benchmark_items["item_id"].to_numpy()
predictions = item_ids[indices]
local_queries.loc[0, "search_query"]
benchmark_items.iloc[indices[0]][["item_id", "item_title_raw"]].head(10)
```

```python
def recall_at_50(predicted_items, relevant_items):
    predicted_set = set(predicted_items)
    relevant_set = set(relevant_items)

    return len(predicted_set & relevant_set) / len(relevant_set)

recall_scores = [
    recall_at_50(predicted, relevant)
    for predicted, relevant in zip(
        predictions,
        local_queries["relevant_item_ids"],
    )
]

local_recall = np.mean(recall_scores)
print(f"Recall@50 = {local_recall}")
```

## 5. Гибрид локального и глобального поиска

Для каждого запроса берём до 40 лучших объявлений из той же локации, затем дополняем
список глобальными результатами до 50. 


```python
local_quota = 40
hybrid_predictions = []

for query_position, query_row in local_queries.iterrows():
    query_vector = query_matrix[query_position]
    query_location_id = query_row["search_location_id"]

    local_mask = (
        benchmark_items["item_location_id"].to_numpy()
        == query_location_id
    )
    local_indices = np.flatnonzero(local_mask)

    local_scores = (
        item_matrix[local_indices] @ query_vector.T
    ).toarray().ravel()

    local_count = min(local_quota, len(local_indices))
    if local_count > 0:
        local_order = np.argsort(local_scores)[-local_count:][::-1]
        selected_indices = list(local_indices[local_order])
    else:
        selected_indices = []

    # Глобальный канал страхует случаи, когда исполнитель находится в другой локации.
    for item_index in global_top_indices[query_position]:
        if item_index not in selected_indices:
            selected_indices.append(item_index)
        if len(selected_indices) == 50:
            break

    predicted_item_ids = (
        benchmark_items.iloc[selected_indices]["item_id"].tolist()
    )
    hybrid_predictions.append(predicted_item_ids)
```

```python
hybrid_recall_scores = [
    recall_at_50(predicted, relevant)
    for predicted, relevant in zip(
        hybrid_predictions,
        local_queries["relevant_item_ids"],
    )
]
np.mean(hybrid_recall_scores)
```

## 6. Параметры запроса и объявления

`search_infm_params_text` содержит ограничения пользователя, а
`item_infm_params_text` — структурированные свойства объявления. Добавляем их к
текстам.


```python
item_texts_with_params = (
    benchmark_items["item_title_raw"].fillna("").astype(str)
    + " "
    + benchmark_items["item_infm_params_text"].fillna("").astype(str)
)

query_texts_with_params = (
    local_queries["search_query"].fillna("").astype(str)
    + " "
    + local_queries["search_infm_params_text"].fillna("").astype(str)
)
```

```python
vectorizer_with_params = TfidfVectorizer(
    lowercase=True,
)
item_matrix_with_params = vectorizer_with_params.fit_transform(
    item_texts_with_params
)

query_matrix_with_params = vectorizer_with_params.transform(
    query_texts_with_params
)
item_matrix_with_params.shape, query_matrix_with_params.shape
```

```python
search_index_with_params = NearestNeighbors(
    n_neighbors=50,
    metric="cosine",
    algorithm="brute",
    n_jobs=-1,
)

search_index_with_params.fit(item_matrix_with_params)
```

```python
distances_with_params, indices_with_params = (
    search_index_with_params.kneighbors(query_matrix_with_params)
)

hybrid_predictions_with_params = []

for query_number, query_row in local_queries.iterrows():
    location = query_row["search_location_id"]

    local_indices = np.asarray(
        location_to_indices.get(location, []),
        dtype=int,
    )

    local_scores = (
        item_matrix_with_params[local_indices]
        @ query_matrix_with_params[query_number].T
    ).toarray().ravel()

    positive_positions = np.flatnonzero(local_scores > 0)

    top_local_positions = positive_positions[
        np.argsort(local_scores[positive_positions])[-40:][::-1]
    ]

    top_local_indices = local_indices[top_local_positions]

    combined_indices = list(top_local_indices)
    seen_indices = set(combined_indices)

    for item_index in indices_with_params[query_number]:
        if item_index not in seen_indices:
            combined_indices.append(item_index)
            seen_indices.add(item_index)

        if len(combined_indices) == 50:
            break

    hybrid_predictions_with_params.append(
        item_ids[combined_indices]
    )
```

```python
recall_with_params = [
    recall_at_50(predicted, relevant)
    for predicted, relevant in zip(
        hybrid_predictions_with_params,
        local_queries["relevant_item_ids"],
    )
]

np.mean(recall_with_params)
```

```python
np.mean(np.array(recall_with_params) > 0)
```

### Результат после добавления параметров

|---|---:|
| Recall@50 | 0.6187 |
| HitRate@50 | 0.6305 |

Параметры дают заметный прирост относительно поиска только по заголовкам. Значит,
это отдельный полезный текстовый канал, а не вспомогательная колонка.


## 7. Описание как дополнительный канал

Описание содержит больше деталей, но оно длинное и шумное. Поэтому не заменяем им
основную выдачу: берём 40 кандидатов по заголовку и параметрам и добавляем до 10
кандидатов из отдельного TF-IDF индекса описаний.


```python
failed_indices = np.flatnonzero(
    np.array(recall_with_params) == 0
)

rng = np.random.default_rng(42)
failed_index = int(rng.choice(failed_indices))

local_queries.loc[
    failed_index,
    search_columns,
]
```

```python
relevant_ids = local_queries.loc[
    failed_index,
    "relevant_item_ids",
]

benchmark_items[
    benchmark_items["item_id"].isin(relevant_ids)
][
    [
        "item_id",
        "item_title_raw",
        "item_infm_params_text",
        "item_description_raw",
        "item_location_id",
        "item_category_id",
    ]
]
```

```python
benchmark_items.iloc[
    indices_with_params[failed_index]
][
    ["item_id", "item_title_raw", "item_location_id"]
].head(10)
```

На примере «прокат бмв» релевантное объявление использует слово «аренда». Обычный
словный TF-IDF не понимает, что «прокат» и «аренда» близки по смыслу, поэтому его
оценка для этой пары равна нулю.


```python
correct_item_id = local_queries.loc[
    failed_index,
    "relevant_item_ids",
][0]

correct_position = benchmark_items.index[
    benchmark_items["item_id"] == correct_item_id
][0]
```

```python
description_texts = (
    benchmark_items["item_description_raw"]
    .fillna("")
    .astype(str)
)

description_vectorizer = TfidfVectorizer(
    lowercase=True,
    min_df=2,
    max_features=200_000,
    dtype=np.float32,
)
description_matrix = description_vectorizer.fit_transform(
    description_texts
)

description_query_matrix = description_vectorizer.transform(
    query_texts_with_params
)
description_score = (
    description_matrix[correct_position]
    @ description_query_matrix[failed_index].T
).toarray()[0, 0]

description_score

```

```python
description_search_index = NearestNeighbors(
    n_neighbors=50,
    metric="cosine",
    algorithm="brute",
    n_jobs=-1,
)

description_search_index.fit(description_matrix)
```

```python
description_distances, description_indices = (
    description_search_index.kneighbors(
        description_query_matrix
    )
)
description_indices.shape
```

```python
description_predictions = item_ids[description_indices]
combined_predictions = []

for primary, description in zip(
    hybrid_predictions_with_params,
    description_predictions,
):
    combined = list(primary[:40])
    seen = set(combined)

    for item_id in description:
        if item_id not in seen:
            combined.append(item_id)
            seen.add(item_id)

        if len(combined) == 50:
            break

    # На случай большого числа повторов заполняем остаток
    # кандидатами основного канала.
    if len(combined) < 50:
        for item_id in primary[40:]:
            if item_id not in seen:
                combined.append(item_id)
                seen.add(item_id)

            if len(combined) == 50:
                break

    combined_predictions.append(combined)

combined_recall_scores = [
    recall_at_50(predicted, relevant)
    for predicted, relevant in zip(
        combined_predictions,
        local_queries["relevant_item_ids"],
    )
]

np.mean(combined_recall_scores)
```

## 8. Проба семантического поиска E5

Мультиязычная модель `intfloat/multilingual-e5-small` кодирует смысл текста в плотный
вектор. Для E5 запросы получают префикс `query:`, а объявления — `passage:`.
На этом этапе проверяем подход качественно на разобранной ошибке, без полного расчёта метрики.


```python
import os

# На некоторых машинах переменная указывает на несуществующий файл SSL-логов.
os.environ.pop("SSLKEYLOGFILE", None)

from sentence_transformers import SentenceTransformer
```

```python
semantic_model = SentenceTransformer(
    "intfloat/multilingual-e5-small"
)
```

```python
semantic_model.device
```

```python
test_query = "query: прокат бмв"

test_document = (
    "passage: "
    + str(benchmark_items.loc[correct_position, "item_title_raw"])
    + " "
    + str(benchmark_items.loc[correct_position, "item_infm_params_text"])
    + " "
    + str(benchmark_items.loc[correct_position, "item_description_raw"])
)
```

```python
embeddings = semantic_model.encode(
    [test_query, test_document],
    normalize_embeddings=True,
)
semantic_score = embeddings[0] @ embeddings[1]
semantic_score
```

```python
candidate_positions = [
    correct_position,
    *indices_with_params[failed_index][:10],
]

candidate_positions = list(dict.fromkeys(candidate_positions))

candidate_rows = benchmark_items.iloc[candidate_positions].copy()
candidate_texts = (
    "passage: "
    + candidate_rows["item_title_raw"].fillna("").astype(str)
    + " "
    + candidate_rows["item_infm_params_text"].fillna("").astype(str)
    + " "
    + candidate_rows["item_description_raw"].fillna("").astype(str)

)

query_embedding = semantic_model.encode(
    [test_query],
    normalize_embeddings=True,
)[0]

candidate_embeddings = semantic_model.encode(
    candidate_texts.tolist(),
    normalize_embeddings=True,
)
```

```python
comparison = candidate_rows[
    ["item_id", "item_title_raw"]
].copy()

comparison["semantic_score"] = (
    candidate_embeddings @ query_embedding
)

comparison["is_relevant"] = comparison["item_id"].isin(
    relevant_ids
)

comparison.sort_values(
    "semantic_score",
    ascending=False,
)
```

```python
test_location = local_queries.loc[
    failed_index,
    "search_location_id",
]

location_positions = np.asarray(
    location_to_indices[test_location]
)

location_items = benchmark_items.iloc[
    location_positions
].copy()


location_documents = (
    "passage: "
    + location_items["item_title_raw"].fillna("").astype(str)
    + " "
    + location_items["item_infm_params_text"].fillna("").astype(str)
    + " "
    + location_items["item_description_raw"]
        .fillna("")
        .astype(str)
        .str[:1000]
)
location_embeddings = semantic_model.encode(
    location_documents.tolist(),
    batch_size=32,
    normalize_embeddings=True,
    show_progress_bar=True,
)

location_scores = location_embeddings @ query_embedding
location_order = np.argsort(-location_scores)

correct_rank = (
    np.where(
        location_positions[location_order]
        == correct_position
    )[0][0]
    + 1
)

correct_rank
```

### Что показал семантический пример

| Наблюдение | Значение |
|---|---:|
| Косинусная близость «прокат бмв» и релевантного объявления | 0.8340 |
| Позиция релевантного объявления среди разобранных кандидатов | 9 |
| Позиция при поиске внутри локации | 13 |

E5 не делает релевантное объявление первым, но возвращает его в кандидатный список,
где TF-IDF давал нулевую близость. Это достаточное основание проверить
семантический канал на всей локальной валидации.

### Итоговые решения для pipeline

1. Использовать несколько retrieval-каналов, а не один общий текст.
2. Сохранять локальную квоту и глобальный fallback.
3. Учитывать категорию как ограничение или сильный признак.
4. Добавить семантический E5-поиск для синонимов и перефразирований.
5. Сравнивать версии на честном разбиении по полному поисковому контексту.

Числа выше относятся к исследовательской выборке и нужны для выбора направления.
Окончательные метрики необходимо пересчитать на всей выборке.
